```python
import json
from fastavro import reader

with open("users.avro", "rb") as src, \
     open("users.jsonl", "w") as dst:

    for record in reader(src):
        dst.write(json.dumps(record, ensure_ascii=False) + "\n")
```

```python
%%bash
ls -la *.jsonl
```

```python
import pandas as pd

df = pd.read_json(
    "users.jsonl",
    lines=True
)

df.to_csv(
    "users.csv",
    index=False
)
```
