## CSVをSQLiteにインポート
CSVをSQLiteにインポートしてクエリを実行する。

```python
%%bash
set -e
sqlite3 -bail demo.db <<'SQL'
BEGIN;
DROP TABLE IF EXISTS users_csv;
.import --csv users.csv users_csv
COMMIT;
SQL
```

```python
%load_ext sql
%sql sqlite:///demo.db
```

```python
%%sql
SELECT id, name FROM users_csv ORDER BY id;
```

```python
%sql --close sqlite:///demo.db
# Close the connection to the SQLite database
```
