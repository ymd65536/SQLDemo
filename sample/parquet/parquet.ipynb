## CSVからParquetに変換する 

```python
import pandas as pd

df = pd.read_csv("users.csv")
df.to_parquet("users.parquet", index=False)
```

```python
%%bash
ls -la users.parquet
```

```python
import pandas as pd

df = pd.read_parquet("users.parquet")

df.to_json(
    "users.jsonl",
    orient="records",
    lines=True
)
```

```python
import json
from fastavro import writer, parse_schema

schema = {
    "type": "record",
    "name": "Employee",
    "fields": [
        {"name": "id", "type": "int"},
        {"name": "name", "type": "string"},
    ]
}

records = []

with open("users.jsonl", "r") as f:
    for line in f:
        records.append(json.loads(line))

with open("users.avro", "wb") as out:
    writer(out, parse_schema(schema), records)
```
