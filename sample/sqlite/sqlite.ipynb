```python
%%bash
sudo apt-get install -y sqlite3 > /dev/null
echo "sqlite3 installation completed."
```

```python
# create sqlite database and connect to it
%load_ext sql
%sql sqlite:///demo.db
```

```python
%%sql
CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL
);

INSERT INTO users (name) VALUES ('Alice'), ('Bob');
```

```python
%%sql
SELECT
  id,
  name
FROM users;
```

```python
%%bash
set -e
sqlite3 -header -csv demo.db "SELECT * FROM users ORDER BY id;" > users.csv
cat users.csv
```

```python
%%bash
# copy users.csv from sample/sqlite to sample/csv
cp users.csv ../csv
echo "移動先のディレクトリの内容"
ls ../csv
```

```python
%sql --close sqlite:///demo.db
# Close the connection to the SQLite database
```
